RESHAPE , STACKING , SQEEZING , UNSQEEZING
1 - RESHAPING : RESHAPING TENSORS TO A DEFINE SHAPE
2 - VIEW : VIEW THE TENSOTR OF THE CURRENT SHAPE BUT KEEP THE ORIGNAL TO THE MEMORY
3 - STACKING : COMBINE MULTIPLE TENSORS ON TOP OTHER VSTACK VERTIVAL STACK OR HSTACK HORIZONTAL STACK
4 - SQEEZE : REMOVES ALL DIMENSIONS 1 FROM TENSORS
5 - UNSQEEZE : ADD A 1 DIMENSIONS TO A TARGET TENSOR
6 - PERMUTE : RETURN A VIEW FROM INPUT WITH PERMUTED DIMENSIONS (SWAPPED) IN A CERTAIN WAY

```python
shapearray = torch.arange(1. , 10.)
shapearray , shapearray.shape
```

```python
reshapedtensor = shapearray.reshape(1 , 9)
reshapedtensor , reshapedtensor.shape
```

```python
reshapedtensor = shapearray.reshape(9 ,1)
reshapedtensor , reshapedtensor.shape
```
