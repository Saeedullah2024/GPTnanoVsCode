RESHAPE , STACKING , SQEEZING , UNSQEEZING
1 - RESHAPING : RESHAPING TENSORS TO A DEFINE SHAPE
2 - VIEW : VIEW THE TENSOTR OF THE CURRENT SHAPE BUT KEEP THE ORIGNAL TO THE MEMORY
3 - STACKING : COMBINE MULTIPLE TENSORS ON TOP OTHER VSTACK VERTIVAL STACK OR HSTACK HORIZONTAL STACK
4 - SQEEZE : REMOVES ALL DIMENSIONS 1 FROM TENSORS
5 - UNSQEEZE : ADD A 1 DIMENSIONS TO A TARGET TENSOR
6 - PERMUTE : RETURN A VIEW FROM INPUT WITH PERMUTED DIMENSIONS (SWAPPED) IN A CERTAIN WAY

```python
shapearray = torch.arange(1. , 13.)
shapearray , shapearray.shape
```

```python
reshapedtensor = shapearray.reshape(1 , 9)
reshapedtensor , reshapedtensor.shape
```

```python
reshapedtensor = shapearray.reshape(9 ,1)
reshapedtensor , reshapedtensor.shape
```

```python
reshapedtensor = shapearray.reshape(4 , 3)
reshapedtensor , reshapedtensor.shape
```

```python
z = shapearray.view(3 ,4 ) #Now z have the same memory address of the shapearray
z , z.shape
```

```python
z[:]
```

```python
z[0:1: , 0:1] = 100
z , shapearray
```

NOW ITS TIME FOR STACKING = HORIZONTAL AND VERTICAL STACKING

```python
stackingarray = torch.arange(1 , 7)
stackingarray , stackingarray.ndim
```

```python
x = stackingarray.reshape(1 , 6)
x , x.shape , x.ndim
```

```python
hstack = torch.stack([x , x , x] , dim=0)
hstack , hstack.shape , hstack.ndim
```

```python
hstack = torch.stack([x , x , x] , dim=1)
hstack , hstack.shape , hstack.ndim
```

```python
stackingmatrix = torch.rand(3 ,5)
stackingmatrix , stackingmatrix.shape , stackingmatrix.ndim
```

```python
stackingmatrixStacked = torch.stack([stackingmatrix , stackingmatrix] , dim=0)
stackingmatrixStacked , stackingmatrixStacked.shape , stackingmatrixStacked.ndim
#torch.Size([2, 3, 5]), 2 form with pair of 3
```

```python
stackingmatrixStacked = torch.stack([stackingmatrix , stackingmatrix] , dim=1)
stackingmatrixStacked , stackingmatrixStacked.shape , stackingmatrixStacked.ndim
#torch.Size([3, 2, 5]), 3 form with pair of 2
```
