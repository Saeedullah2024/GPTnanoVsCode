NUMPY TO TENSORS

```python
#This is not like the View or Permutate. it doesnot share the address and the values in one change doesnot effect the other
array = np.arange(1.0 , 9.0) #-> this will give a data type of float 64
tensors = torch.from_numpy(array).type(torch.float32) #because the array is float 64 the tensor will also be float 64 we have to convert it
array.dtype , tensors.dtype 
```

REPRODUCIBILITY (TRYPING TO TAKEN OUT RANDOM FROM RANDOM)

```python
# To reduce the randomness in the neural network the pytorch comes with the concept of the random seed. 
# The random seed is a number that is used to initialize the random number generator. 
# The random number generator is used to generate random numbers. 
# The random numbers are used to initialize the weights of the neural network. 
# The random seed is used to ensure that the random numbers generated are the same every time the code is run. 
# This is useful for debugging and for reproducing results.
```

```python
import torch
import numpy as np
```

```python
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
tensor1 = torch.rand(3 , 3)

torch.manual_seed(RANDOM_SEED)
tensor2 = torch.rand(3 , 3)
```
