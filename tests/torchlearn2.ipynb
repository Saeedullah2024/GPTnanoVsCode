Two Ways of doing Multiplication in Neural network and Deep learing /n
1- Element - wise multiplication
2- Matrix Multiplication - Dot Product

```python
#Element Wise multiplication
print(f"{float_32_tensor} || {float_64_tensor}")
print(f"{float_64_tensor * float_32_tensor}")
```

```python
#For Dot multiplication we use torch.matmul
dotproduct = torch.matmul(float_64_tensor , float_32_tensor)
dotproduct
```
