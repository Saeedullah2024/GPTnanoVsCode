```python
import torch
```

```python
#torch.sqeeze = Remove all single dimension from the target tensor
```

```python
tesnor = torch.rand(1 , 9) 
#It is to create number from 0 to 1 with the mean of 0 and std of 2.8
#torch.randn = It is use to create number from negattive to psotive and mean is 0 and standard deviartuob  is aboyt 1
tesnor , tesnor.shape , tesnor.ndim
```

```python
print(f"The tensor is {tesnor} \nThe dimension is {tesnor.ndim} \nThe shape is {tesnor.shape}")
```

```python
print("After the Sqeeze")
newtensor = tesnor.squeeze()
print(f"The previous tenosr become {newtensor} \nThe shape of the Tensor is {newtensor.shape} \nThe size of the tensor is {newtensor.ndim}")
```

```python
#unsqeeze() - This will add the single dimension into the targeted tensor
```

```python
newtensor
```

```python
tensorunsqeeze = newtensor.unsqueeze(0) # 0 for the adding at 0 position
print(f"The previous tensor was {newtensor} \nThe size was {newtensor.size} \nThe shape is {newtensor.shape}")
print("\n")
print(f"The new tensor with unsqeeze is {tensorunsqeeze} \nThe size is {tensorunsqeeze.shape} \nThe dimension is {tensorunsqeeze.ndim}")
```
